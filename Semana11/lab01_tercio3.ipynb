## 1) Descripción del problema **[1 pts]**

### 5) Análisis de complejidad **[15 pts]**

#### Left, right, parent
- Justificacion: Operaciones aritméticas básicas para el cálculo de índices en un arreglo, ejecutadas en tiempo constante.
- Mejor y peor caso: O(1)



#### Max Heapify:
- Mejor caso: O(1): El nodo evaluado en la posición i ya posee una prioridad mayor o igual a la de sus hijos, deteniendo la recursión inmediatamente.
- Peor caso: O(log (n)). El valor en el nodo es menor que sus descendientes y debe "hundirse" intercambiándose hasta el nivel más bajo (hojas) del árbol. Dado que el árbol es casi completo, recorre su altura h = log base 2 de (n).



#### Build max heap:
- Mejor caso: O(n): Si el arreglo de entrada ya está organizado internamente como un Max-Heap, las llamadas a max_heapify verifican a los hijos y terminan en O(1) cada una, resultando en un tiempo estrictamente proporcional a la cantidad de nodos.
- Peor caso: O(n). Aunque llama a una función O(log n) repetidas veces, la mayoría de los nodos están en la base del árbol y tienen altura cercana a 0. La sumatoria del trabajo total en todos los niveles está acotada por un tiempo lineal.



#### Maximum:
- Mejor y Peor caso: O(1): Únicamente accede y retorna el valor ubicado en el índice 0 del arreglo. No realiza búsquedas ni modificaciones en la estructura de datos.



#### Extract Max:
- Mejor caso: O(1). Si el montículo tiene solo uno o dos elementos, o si el último elemento movido temporalmente a la raíz ya cumple la propiedad del montículo sin necesidad de hundirse.
- Peor caso: O(log n). Se extrae la raíz y se reemplaza por el último elemento. La llamada subsecuente a max_heapify debe descender este elemento recorriendo toda la altura del árbol.



#### Inset:
- Mejor caso: O(1). La nueva tarea se inserta al final del arreglo y su prioridad es menor que la de su nodo padre, por lo que el bucle de ascenso se rompe de inmediato sin ejecutar intercambios.
- Peor caso: O(log n). La tarea insertada posee la máxima prioridad global, forzando al algoritmo a realizar comparaciones e intercambios nivel por nivel hasta moverla desde una hoja hasta la raíz.



#### Heapsort:
- Mejor caso O(n log n). Todo elemento debe extraerse del montículo para ser ordenado al final del arreglo. Incluso con elementos repetidos o parcialmente ordenados, reacomodar la estructura para mantener sus propiedades mecánicas toma un tiempo logarítmico para cada uno de los n elementos.
- Peor caso: O(n log n). Consiste en llamar a build_max_heap en tiempo O(n) seguido de n-1 extracciones donde cada reorganización cuesta O(log n).

```python
def left(i):
    return 2 * i + 1

def right(i):
    return 2 * i + 2

def parent(i):
    return (i - 1) // 2

def max_heapify(arr, n, i):
    largest = i
    l = left(i)
    r = right(i)

    if l < n and arr[l]["priority"] > arr[largest]["priority"]:
        largest = l
    if r < n and arr[r]["priority"] > arr[largest]["priority"]:
        largest = r

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        max_heapify(arr, n, largest)

def build_max_heap(tasks):
    heap = list(tasks)
    n = len(heap)
    for i in range(n // 2 - 1, -1, -1):
        max_heapify(heap, n, i)
    return heap

def maximum(heap):
    if not heap:
        return None
    return heap[0]

def extract_max(heap):
    n = len(heap)
    if n == 0:
        return None
    if n == 1:
        return heap.pop()
    
    root = heap[0]
    heap[0] = heap.pop()
    max_heapify(heap, len(heap), 0)
    return root

def insert(heap, task):
    heap.append(task)
    i = len(heap) - 1
    while i > 0 and heap[parent(i)]["priority"] < heap[i]["priority"]:
        heap[i], heap[parent(i)] = heap[parent(i)], heap[i]
        i = parent(i)

def heapsort(heap):
    arr = list(heap) 
    n = len(arr)
    for i in range(n - 1, 0, -1):
        arr[i], arr[0] = arr[0], arr[i]
        max_heapify(arr, i, 0)
    return arr

def print_tree(heap):
    priorities = [t["priority"] for t in heap]
    n = len(priorities)
    if n == 0:
        return
    print(f"      {priorities[0]}")
    if n > 1:
        c1 = priorities[1]
        c2 = priorities[2] if n > 2 else ""
        print("    /   \\")
        print(f"   {c1}     {c2}")
    if n > 3:
        c3 = priorities[3]
        c4 = priorities[4] if n > 4 else ""
        c5 = priorities[5] if n > 5 else ""
        c6 = priorities[6] if n > 6 else ""
        print("  / \\   / \\")
        print(f" {c3}   {c4} {c5}   {c6}")


print("--- Punto 1. Registrar las tareas ---")
while True:
    try:
        n = int(input("Ingrese la cantidad n de tareas: "))
        if n > 0:
            break
        print("Debe ingresar un número mayor a 0.")
    except ValueError:
        print("Error: Ingrese un número entero válido.")

tareas_originales = []
for i in range(n):
    print(f"\nIngresando Tarea {i+1}:")
    codigo = input("Código (ej. T01): ")
    nombre = input("Nombre de la tarea: ")
    while True:
        try:
            prioridad = int(input("Prioridad numérica: "))
            break
        except ValueError:
            print("Error: La prioridad debe ser un número entero.")
    tareas_originales.append({"code": codigo, "name": nombre, "priority": prioridad})

prioridades_originales = [t["priority"] for t in tareas_originales]
print(f"\nLista original de prioridades: {prioridades_originales}")

print("\n--- Punto 2. Construir el Max-Heap ---")
heap = build_max_heap(tareas_originales)
print(f"Lista original: {prioridades_originales} Max-Heap: {[t['priority'] for t in heap]}")
print("Árbol:")
print_tree(heap)

print("\n--- Punto 3A. Consultar la tarea de mayor prioridad ---")
tarea_max = maximum(heap)
print(f"Tarea de mayor prioridad: {tarea_max['code']} - {tarea_max['name']} Prioridad: {tarea_max['priority']}")

print("\n--- Punto 3B. Procesar la tarea de mayor prioridad ---")
tarea_procesada = extract_max(heap)
print(f"Tarea procesada: {tarea_procesada['code']} - {tarea_procesada['name']}")
print(f"Nuevo Heap: {[t['priority'] for t in heap]}")

print("\n--- Punto 3C. Insertar una nueva tarea ---")
print("Ingrese los datos de la nueva tarea a insertar:")
codigo_nuevo = input("Código: ")
nombre_nuevo = input("Nombre: ")
while True:
    try:
        prioridad_nueva = int(input("Prioridad numérica: "))
        break
    except ValueError:
        print("Error: La prioridad debe ser un número entero.")
        
insert(heap, {"code": codigo_nuevo, "name": nombre_nuevo, "priority": prioridad_nueva})
print(f"Nuevo Heap tras inserción: {[t['priority'] for t in heap]}")

print("\n--- Punto 4. HeapSort ---")
lista_ordenada = heapsort(heap)
print(f"Lista original (histórico): {prioridades_originales}")
print(f"Max-Heap actual: {[t['priority'] for t in heap]}")
print(f"Lista ordenada: {[t['priority'] for t in lista_ordenada]}")


```
